Same synthetic setup as Notebook 02 (K=40, T=1, r_true=0.05, sigma_true=0.20,

Only change from the original settings: sigma_raw gets its own learning

```python
from inverse_pinn_diff_lr_training import train_one_run_diff_lr
```

```python
lr_network = 1e-3
lr_r = 1e-3
lr_sigma = 2e-3
```

### Section 1: Fixed seed=42 testing

```python
result_42 = train_one_run_diff_lr(
    run_number=1,
    lr_network=lr_network,
    lr_r=lr_r,
    lr_sigma=lr_sigma,
    seed=42,
    checkpoint_prefix="european_inverse_diff_lr_seed42"
)

result_42
```

### Section 2: Run 10 times (no fixed seed) & take average

```python
import pandas as pd

results = []

for run in range(1, 11):

    result = train_one_run_diff_lr(
        run_number=run,
        lr_network=1e-3,
        lr_r=1e-3,
        lr_sigma=2e-3,
        seed=None,
        checkpoint_prefix="european_inverse_diff_lr_checkpoint"
    )

    results.append(result)

    pd.DataFrame(results).to_csv(
        "european_inverse_diff_lr_10run_results.csv",
        index=False
    )
```

```python
results_df = pd.DataFrame(results)

summary_df = results_df[
    [
        "r_estimate",
        "sigma_estimate",
        "r_error_percent",
        "sigma_error_percent",
        "price_mse",
        "price_correlation",
        "runtime_minutes",
    ]
].agg(["mean", "std"])

summary_df
```
