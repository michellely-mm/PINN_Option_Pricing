# 05 — Forward PINN for the European Call

This notebook trains the **forward PINN** using the same European-call domain and neural-network architecture as the inverse experiment.

- $r=0.05$ and $\sigma=0.20$ are **known and fixed**.
- Only the neural-network weights are trained.
- Training uses PDE collocation points + terminal points + boundary points.
- The $100\times100$ mesh is used **only after training** for testing.
- The trained model is saved to `european_forward_visualization_checkpoint.pt`.

```python
import importlib
import torch
import forward_pinn_model
import forward_pinn_training

importlib.reload(forward_pinn_model)
importlib.reload(forward_pinn_training)

from forward_pinn_model import ForwardPINN
from forward_pinn_training import train_one_run
```

```python
# Train one forward-PINN run.
# With 30,000 epochs on CPU this can take several minutes.

result, forward_model, forward_history = train_one_run(
    run_number=1,
    print_every=5000,
    return_model=True,
    track_history=True,
    seed=42
)

result
```

```python
# Save the trained forward PINN for visualization in Notebook 03.

checkpoint = {
    "model_state_dict": forward_model.state_dict(),
    "history": forward_history,
    "result": result,
    "t_min": forward_pinn_training.t_min,
    "t_max": forward_pinn_training.t_max,
    "S_min": forward_pinn_training.S_min,
    "S_max": forward_pinn_training.S_max,
    "K": forward_pinn_training.K,
    "T": forward_pinn_training.T,
    "r_true": forward_pinn_training.r_true,
    "sigma_true": forward_pinn_training.sigma_true,
    "hidden_width": forward_pinn_training.hidden_width,
    "hidden_layers": forward_pinn_training.hidden_layers,
    "num_epochs": forward_pinn_training.num_epochs,
    "N_f": forward_pinn_training.N_f,
    "N_terminal": forward_pinn_training.N_terminal,
    "N_boundary": forward_pinn_training.N_boundary
}

torch.save(
    checkpoint,
    "european_forward_visualization_checkpoint.pt"
)
```
