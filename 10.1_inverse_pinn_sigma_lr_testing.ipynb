```python
import pandas as pd

results = []

for run in range(1, 11):

    result = train_one_run_diff_lr(
        run_number=run,
        lr_network=1e-3,
        lr_r=1e-3,
        lr_sigma=3e-4,
        seed=None,
        checkpoint_prefix="european_inverse_diff_lr_checkpoint"
    )

    results.append(result)

    pd.DataFrame(results).to_csv(
        "european_inverse_diff_lr_10run_results.csv",
        index=False
    )
```

```python
results_df = pd.DataFrame(results)

summary_df = results_df[
    [
        "r_estimate",
        "sigma_estimate",
        "r_error_percent",
        "sigma_error_percent",
        "price_mse",
        "price_correlation",
        "runtime_minutes",
    ]
].agg(["mean", "std"])

summary_df
```
